```python
def expt(b, n):
	total = 1
	for i in range(n):
		total *= b
	return total
print(expt(6, 3))
```

```python
def fast_expt(b,n):
	print(b, n)
	if n==0:
		return 1
	else:
		if n%2 == 0: # 짝수
			return square(fast_expt(b, n//2))
		else:
			return b*fast_expt(b, n-1)

def square(b):
	return b*b

print(fast_expt(2, 5))
```
